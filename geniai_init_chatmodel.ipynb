### LangChain Model Initialization using init_chat_model()

##### 1. Introduction to init_chat_model()
- In LangChain, init_chat_model() is a function used to initialize different AI chat models through a common interface.

- Instead of directly using provider-specific SDK methods, we can use LangChain's init_chat_model() to connect with different AI model providers such as Google Gemini, OpenAI and Groq.

- Example: We can initialize Gemini, GPT and Llama models using the same LangChain function.

##### 2. Why Do We Need init_chat_model()?
- To initialize different AI models using a common interface.
- To reduce provider-specific coding.
- To switch between different AI models easily.
- To integrate AI models with LangChain components.
- To build chains, agents and RAG applications.
- To simplify model integration in Generative AI applications.

### 5. Basic Syntax of init_chat_model()

```python
from langchain.chat_models import init_chat_model

model = init_chat_model(
    "model-name",
    model_provider="provider-name"
)

# Explanation:
# - init_chat_model() – Initializes an AI chat model.
# - model-name – Specifies which AI model we want to use.
# - model_provider – Specifies the provider of the AI model.
# - model – Stores the initialized chat model object.
```

### Initialize Gemini Model

```python
from dotenv import load_dotenv
from langchain.chat_models import init_chat_model

load_dotenv()

model = init_chat_model(
    "gemini-3.5-flash",
    model_provider="google_genai"
)

response = model.invoke("What is Python in 2 lines?")

print(response.content)
```

### Chat Model Calling with Fallback
- You can use Fallback Model Calling in LangChain. If the primary model is unavailable or fails, the application automatically tries another model.
- Primary model: Gemini
- Fallback model: OpenAI
- If Gemini fails, OpenAI generates the response.

```python
from dotenv import load_dotenv
from langchain.chat_models import init_chat_model

load_dotenv()

# Primary Model
gemini = init_chat_model(
    "gemini-2.5-flash",
    model_provider="google_genai"
)

# First Fallback
openai = init_chat_model(
    "gpt-5.5",
    model_provider="openai"
)

# Second Fallback
groq = init_chat_model(
    "qwen/qwen3.8-27b",
    model_provider="groq"
)

# Configure Multiple Fallbacks
model = gemini.with_fallbacks([openai, groq])

# Model Calling
try:
    response = model.invoke("What is Python in one line?")
    print(response.content)

except Exception as e:
    print("All models failed:", e)
```
