### Model Calling in Generative AI
- Model Calling is the process of sending a prompt (input) from a Python application to an AI model through an API and receiving a generated response.

- In Generative AI, we use model calling to interact with different Large Language Models (LLMs), such as Gemini, GPT and Llama.

#### Why Do We Need Model Calling?
- To interact with AI models using Python applications.
- To generate human-like text responses.
- To build AI-powered applications.
- To automate tasks using LLMs.
- To integrate AI capabilities into real-world projects.

##### genai.Client() – Creates a client to communicate with Google's Gemini API.
##### client.models.generate_content() – Directly sends a prompt to the Gemini model.
##### model – Specifies which Gemini model to use.
##### contents – Contains the input prompt.
##### response.text – Retrieves the generated response.

```python
from google import genai

client = genai.Client(api_key="AQ.Ab8RN6KpOIdDol7d0ez2aobCckmIxHEf4_xhyWQhTIvJCzyBMQ")

response = client.models.generate_content(
    model="gemini-3.5-flash",
    contents="What is Python in 2 lines?"
)

print(response.text)
```

## Example 2: With a user-defined function for gemini

##### Instead of writing the API-calling code repeatedly, we create one function and reuse it with different prompts.

```python
from google import genai

client = genai.Client(api_key="AQ.Ab8RN6KpOIdDol7d0ez2aobCckmIxHEf4_xhyWQhTIvJCzyBMQ")

def ask_gemini(prompt):
    response = client.models.generate_content(
        model="gemini-3.5-flash",
        contents=prompt
    )
    return response.text

result = ask_gemini("What is Python in 2 lines?")
print(result)


# print(ask_gemini("What is AWS?"))
# print(ask_gemini("What is Docker?"))
# print(ask_gemini("What is Kubernetes?"))
```

## 2. OpenAI model calling
#### SDK: openai

```python
%pip install openai
```

#### Example 1: Without a user-defined function

##### from openai import OpenAI – Imports OpenAI SDK.
######  OpenAI() – Creates the OpenAI client.
######  client.responses.create() – Sends a request to the OpenAI model.
######  model – Specifies the GPT model.
######  input – Sends the prompt.
######  response.output_text – Extracts the generated text.

```python
from openai import OpenAI

client = OpenAI(api_key="YOUR_OPENAI_API_KEY")

response = client.responses.create(
    model="gpt-5.5",
    input="What is Python in 2 lines?"
)

print(response.output_text)

```

### Example 2: With a user-defined function

```python
from openai import OpenAI

client = OpenAI(api_key="YOUR_OPENAI_API_KEY")

def ask_openai(prompt):

    response = client.responses.create(
        model="gpt-5.5",
        input=prompt
    )

    return response.output_text


result = ask_openai("What is Python in 2 lines?")

print(result)

### Calling the function multiple times:

# print(ask_openai("Explain AWS"))
# print(ask_openai("Explain DevOps"))
# print(ask_openai("Explain Generative AI"))
```

## 3. Groq model calling
##### SDK: Grooq

### Example 1: Without a user-defined function

```python
from groq import Groq

client = Groq(api_key="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")

response = client.chat.completions.create(
    model="qwen/qwen3.8-27b",
    messages=[
        {
            "role": "user",
            "content": "What is Python in 2 lines?"
        }
    ]
)

print(response.choices[0].message.content)



# Explanation:
# - from groq import Groq – Imports Groq SDK.
# - Groq() – Creates the Groq client.
# - client.chat.completions.create() – Sends a chat completion request.
# - model – Selects the model hosted by Groq.
# - messages – Contains conversation messages.
# - role: user – Represents the user's message.
# - content – Contains the actual prompt.
# - response.choices[0].message.content – Extracts the model's answer.
```

### Example 2: With a user-defined function

```python
from groq import Groq

client = Groq(api_key="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")

def ask_groq(prompt):

    response = client.chat.completions.create(
        model="qwen/qwen3.8-27b",
        messages=[
            {
                "role": "user",
                "content": prompt
            }
        ]
    )

    return response.choices[0].message.content


result = ask_groq("What is Python in 2 lines?")

print(result)


# Calling multiple times:
    # print(ask_groq("Explain AWS"))
    # print(ask_groq("Explain Docker"))
    # print(ask_groq("Explain Kubernetes"))
```
