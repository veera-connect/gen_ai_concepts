- In LangChain, we use different message types to communicate with an AI model. Instead of sending only a simple string prompt, we can send a conversation containing multiple messages.

- LangChain provides three important message types.
- Message	Purpose	Meaning
    - SystemMessage	   **Defines AI behaviour	Instructions to AI**
    - HumanMessage	   **Sends user input	Question from user**
    - AIMessage	       **Represents AI output	Previous AI response**


##### Simple classroom explanation:
    - SystemMessage = Tell AI how to behave.
    - HumanMessage = Tell AI what to do.
    - AIMessage = AI tells us the answer

### Understanding with a simple conversation
##### Imagine you are interacting with an AI teacher.
    - SystemMessage: You are a Python teacher.
    - HumanMessage: What is Python?
    - AIMessage: Python is a programming language.
    - HumanMessage: Give me one example.
- The AI receives the complete conversation and uses the previous messages as context.
- Important: AIMessage does not call the AI model. It represents an existing AI response in the conversation history.

### 4. Import Required Libraries

```python
from dotenv import load_dotenv

from langchain_core.messages import (
    SystemMessage,
    HumanMessage,
    AIMessage
)

load_dotenv()


################################################

# Explanation:
# - load_dotenv() loads API keys from the .env file.
# - SystemMessage defines the AI's behaviour.
# - HumanMessage represents the user's question.
# - AIMessage represents a previous AI response.
```

#### 5. Example 1 – Execute Messages Using Gemini AI

```python
from langchain_google_genai import ChatGoogleGenerativeAI

llm = ChatGoogleGenerativeAI(
    model="gemini-2.5-flash",
    temperature=0
)
```

### Create Messages

```python
messages = [
    SystemMessage(
        content="You are a helpful teacher for software engineering."
    ),

    HumanMessage(
        content="What is the difference between a list and tupple in python in two lines?"
    ),

    AIMessage(
        content="A list is mutable, while a tuple is immutable."
    )
]




# Lists are mutable (can be changed after creation), while tuples are immutable (cannot be changed).
# Lists use square brackets `[]`, while tuples use parentheses `()`.
```

### Execute the Messages

```python
response = llm.invoke(messages)

print(response.content)
```

### Execute the Same Messages Using Groq AI

```python
from langchain_groq import ChatGroq

llm = ChatGroq(
    model="qwen/qwen3.8-27b",
    temperature=0
)
```

```python
response = llm.invoke(messages)

print(response.content)
```

## Multiple Questions and Answers in One Conversation

```python
from langchain_groq import ChatGroq

llm = ChatGroq(
    model="qwen/qwen3.8-27b",
    temperature=0
)
```

```python
messages = [
    SystemMessage(content="You are a Python teacher."),

    HumanMessage(content="What is Python?"),

    AIMessage(content="Python is a high-level programming language."),

    HumanMessage(content="What are its main advantages?"),

    AIMessage(content="Python is easy to learn, readable, and has many libraries."),

    HumanMessage(content="Give me three real-world applications.")
]
```

```python
response = llm.invoke(messages)

print(response.content)
```
