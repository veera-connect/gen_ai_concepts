## 1. What is Document Loading?

Document loading means reading data from files such as PDF, TXT, CSV, JSON, and Word.

LangChain converts the loaded information into a common format called a **Document**.

A Document has two main parts:

- `page_content` – the actual text from the file.
- `metadata` – extra information, such as the file name or page number.

Example: When we load a PDF with two pages, the PDF loader usually creates two Documents.

## 2. Document Loaders in LangChain

A document loader is a Python class used to read a particular file type.

Examples:

- PDF – `PyPDFLoader`
- TXT – `TextLoader`
- CSV – `CSVLoader`
- JSON – `JSONLoader`
- Word – `Docx2txtLoader`
- Website – `WebBaseLoader`
- Folder – `DirectoryLoader`

## 3. PDF Loader

### What is PDF Loader?

`PyPDFLoader` is used to read text from PDF files.

Important point: Usually, each PDF page becomes one LangChain Document.

### Step 1: Import PyPDFLoader

```python
from langchain_community.document_loaders import PyPDFLoader
```

### Step 2: Give the PDF file path

```python
file_path = "Data_files/sample.pdf"
```

### Step 3: Create the loader

```python
loader = PyPDFLoader(file_path)
print(loader)
```

### Step 4: Load the PDF

```python
documents = loader.load()
```

### Step 5: Find how many pages were loaded

```python
len(documents)
```

### Step 6: Print the first page content

```python
print(documents[0].page_content)
```

### Step 7: Print metadata

```python
print(documents[0].metadata)
```

### Step 8: Print all pages one by one

```python
for document in documents:
    print(document.page_content)
    print("Metadata:", document.metadata)
    print("--------------------")
```

## 6. JSON Loader

### What is JSONLoader?

JSON files store information in keys and values.

Example: `name` is a key and `Ravi` is its value.

`JSONLoader` uses `jq_schema` to select which JSON information should be loaded.

### Step 1: Import JSONLoader

```python
from langchain_community.document_loaders import JSONLoader
```

### Step 2: Give the JSON file path

### Step 3: Create the loader
Here, `.` means select the complete JSON object.

```python
print(documents[0].metadata)
```

## 9. Directory Loader

### What is DirectoryLoader?

Suppose one folder contains many text files. Instead of loading every file one by one, we can use `DirectoryLoader`.

The `glob="*.txt"` pattern means load all files that end with `.txt`.

### Step 1: Import DirectoryLoader

```python
from langchain_community.document_loaders import DirectoryLoader
```

### Step 2: Create the directory loader
We use `TextLoader` to read each matching text file.

```python
loader = DirectoryLoader(
    "Data_files",
    glob="*.txt",
    loader_cls=TextLoader
)

print(loader)
```

### Step 3: Load all matching text files

```python
documents = loader.load()
```

### Step 4: Check how many Documents were loaded

```python
len(documents)
```

### Step 5: Print all loaded text

```python
for document in documents:
    print(document.page_content)
    print("Metadata:", document.metadata)
    print("--------------------")
```

### Example: Load all PDF files from the folder

```python
pdf_loader = DirectoryLoader(
    "Data_files",
    glob="*.pdf",
    loader_cls=PyPDFLoader
)

pdf_documents = pdf_loader.load()
print("PDF page documents:", len(pdf_documents))
```

## 10. Loading Multiple Documents

### What does this mean?

In a real project, we may have PDF, TXT, CSV, JSON, and Word files together.

We can use the correct loader for each file and combine all the returned Documents into one Python list.

We use `extend()` because each loader returns a list.

### Step 1: Import the loaders individually

```python
from langchain_community.document_loaders import PyPDFLoader
from langchain_community.document_loaders import TextLoader
from langchain_community.document_loaders import CSVLoader
from langchain_community.document_loaders import JSONLoader
from langchain_community.document_loaders import Docx2txtLoader
```

### Step 2: Create an empty list

```python
all_documents = []
```

### Step 3: Load PDF and add its Documents

```python
pdf_loader = PyPDFLoader("Data_files/sample.pdf")
pdf_documents = pdf_loader.load()

all_documents.extend(pdf_documents)
```

### Step 4: Load TXT and add its Documents

```python
text_loader = TextLoader("Data_files/sample.txt")
text_documents = text_loader.load()

all_documents.extend(text_documents)
```

### Step 5: Load CSV and add its Documents

```python
csv_loader = CSVLoader(file_path="Data_files/students.csv")
csv_documents = csv_loader.load()

all_documents.extend(csv_documents)
```

### Step 6: Load JSON and add its Documents

```python
json_loader = JSONLoader(
    file_path="Data_files/student.json",
    jq_schema=".",
    text_content=False
)
json_documents = json_loader.load()

all_documents.extend(json_documents)
```

### Step 7: Load Word and add its Documents

```python
word_loader = Docx2txtLoader("Data_files/notes.docx")
word_documents = word_loader.load()

all_documents.extend(word_documents)
```

### Step 8: Check total Documents

```python
len(all_documents)
```

### Step 9: Print all Documents

```python
for document in all_documents:
    print(document.page_content)
    print("Metadata:", document.metadata)
    print("--------------------")
```
